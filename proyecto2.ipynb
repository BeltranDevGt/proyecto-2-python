```python
import yfinance
```

```python
data = yfinance.Ticker("AAPL").history("6mo")
cierre = data.Close
cierre.plot()
```

## Hacer el analisis solicitado por el gestor
* cotizacion maxima
* cotizacion minima
* cotizacion media de cierre de la accion

```python
maxima = round(cierre.max(), 2)
minima = round(cierre.min(), 2)
valor_medio = round(cierre.mean(), 2)
```

```python
print(maxima)
print(minima)
print(valor_medio)
```
